```python
import pandas as pd
import regex as re
import os
from pathlib import Path
import json
```

```python
def read_json(path):
    text = Path(path).read_text('utf-8')
    return json.loads(text)
```

```python
df = pd.DataFrame(read_json('./artists_songs.json')['artists'])
```

```python
df = df.explode(['songs']).reset_index(drop=True)
```

```python
df['RANK'] = df['songs'].apply(lambda x: x['playlist_position'])
```

```python
df = df.sort_values(['RANK']).reset_index(drop=True)
```

```python
df[df['artist'].apply(lambda x: 'fadl' in x.lower())]
```

```python
df['songs'][0]
```

```python
df = df[df['songs'].apply(lambda x: x['duration_seconds']<420)].reset_index(drop=True)
```

```python
df['tier'].value_counts()
```

```python
cond_01 = df['RANK'] <= 500
cond_02 = df['tier'].apply(lambda x: int(x.split()[-1]) in [1,2])
```

```python
df = df[cond_01 | cond_02].head(500).reset_index(drop=True)
```

```python
df.to_pickle('./top_500_songs.pkl', protocol=4)
```
